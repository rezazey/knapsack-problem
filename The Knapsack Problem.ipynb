```python
def solve_jobs_knapsack(jobs, capacity):
    processes = list(jobs.items())
    n = len(processes)

    
    dp = [[0] * (capacity + 1) for _ in range(n + 1)]
    chosen = [[0] * (capacity + 1) for _ in range(n + 1)]
    

    for i, (_, process) in enumerate(processes, start=1):
        days = process["completion_Day"]
        value = process["value_For_Each_process"]
        limit = process["limit_of_each_process"]

        for c in range(capacity + 1):
            max_quantity = min(limit, c // days)
            
            for quantity in range(max_quantity + 1):
                candidate = dp[i - 1][c - quantity * days] + quantity * value
                if candidate > dp[i][c]:
                    dp[i][c] = candidate
                    chosen[i][c] = quantity

    quantities = {}
    remaining_days = capacity

    for i in range(n, 0, -1):
        key, process = processes[i - 1]
        quantity = chosen[i][remaining_days]
        quantities[key] = quantity
        remaining_days -= quantity * process["completion_Day"]
    
        
    return dp[n][capacity], quantities


best_value, quantities = solve_jobs_knapsack(job_Manufacturing, total_day)

print("Maximum value:", best_value)
print("Process quantities:", quantities)
```

```python


def solve_energy_knapsack(energy, capacity):
    processes = list(energy.items())
    n = len(processes)

    # dp[i][c]: best value using the first i process types within c days
    
    dp = [[0] * (capacity + 1) for _ in range(n + 1)]
    chosen = [[0] * (capacity + 1) for _ in range(n + 1)]
    

    for i, (_, process) in enumerate(processes, start=1):
        value = process["value"]
        limit = process["energyuse"]

        for c in range(capacity + 1):
            max_quantity = c // limit
            
            for quantity in range(max_quantity + 1):
                candidate = dp[i - 1][c - quantity * limit] + quantity * value
                if candidate > dp[i][c]:
                    dp[i][c] = candidate
                    chosen[i][c] = quantity

    quantities = {}
    remaining_days = capacity
    
    for i in range(n, 0, -1):
        key, process = processes[i - 1]
        quantity = chosen[i][remaining_days]
        quantities[key] = quantity
        remaining_days -= quantity * process["energyuse"]
    return dp[n][capacity], quantities


best_value, quantities = solve_energy_knapsack(machine_info, capacity=storage)

print("Maximum value:", best_value)
print("Process quantities:", quantities)
```
